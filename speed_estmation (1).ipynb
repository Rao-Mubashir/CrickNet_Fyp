### Preparing the new dataset for YOLO training


```python
# YOLO directory structure for the new dataset
new_base_path = "/content/datasets/cricket_ball_3"
new_images_train = f"{new_base_path}/train/images"
new_labels_train = f"{new_base_path}/train/labels"
new_images_val = f"{new_base_path}/val/images"
new_labels_val = f"{new_base_path}/val/labels"

# Create folders
os.makedirs(new_images_train, exist_ok=True)
os.makedirs(new_labels_train, exist_ok=True)
os.makedirs(new_images_val, exist_ok=True)
os.makedirs(new_labels_val, exist_ok=True)

print("New YOLO folder structure created!")
```

```python
print("Organizing new dataset...")
new_drive_dataset_path = "/content/drive/MyDrive/crickvision_dataset_2"

for root, dirs, files in os.walk(new_drive_dataset_path):
    for file in files:
        src = os.path.join(root, file)

        # Images
        if file.lower().endswith(('.jpg', '.png', '.jpeg')):
            if "train" in root.lower():
                shutil.copy(src, new_images_train)
            elif "val" in root.lower():
                shutil.copy(src, new_images_val)

        # Labels
        if file.lower().endswith(".txt"):
            if "train" in root.lower():
                shutil.copy(src, new_labels_train)
            elif "val" in root.lower():
                shutil.copy(src, new_labels_val)

print("New dataset organized for YOLO!")
```

```python
new_yaml_path = "/content/datasets/cricket_ball_3.yaml"

new_yaml_content = f"""
path: {new_base_path}
train: train/images
val: val/images

names:
  0: ball
  1: not_ball
"""

with open(new_yaml_path, "w") as f:
    f.write(new_yaml_content)

print("New YAML file created at:", new_yaml_path)
```

```python
# Load the pre-trained model from your Google Drive
model_path_pretrained = "/content/drive/MyDrive/CrickVision/models/best.pt"
model_finetune = YOLO(model_path_pretrained)

# Train the model with the new dataset
# You can adjust epochs, batch size, and learning rate as needed
results_finetune = model_finetune.train(
    data=new_yaml_path,
    imgsz=640,
    epochs=70,
    batch=16,
    lr0=0.0001,
    device=0,
    name="train_v2" # Save results to a new directory
)

print("Fine-tuning finished!")
```

```python
import cv2
from ultralytics import YOLO
import numpy as np
from collections import deque

# ========================= CONFIG =========================
model_path = "/content/drive/MyDrive/CrickVision/models/best_finetuned_on_custom_data.pt"
input_video = "/content/drive/MyDrive/sample.mp4"
output_video = "/content/drive/MyDrive/CrickVision/HAWKEYE_SMOOTH_INSTANT.mp4"

MIN_CONF = 0.25
MIN_SIZE, MAX_SIZE = 4, 100
MAX_JUMP_PER_FRAME = 180
MAX_GAP_FRAMES = 10
TRAIL_MAX_POINTS = 600

# Load your finetuned model
model = YOLO(model_path)
print("Loaded finetuned model: best_finetuned.pt")

cap = cv2.VideoCapture(input_video)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video, fourcc, fps, (w, h))

# Smooth trail storage
trail_points = deque(maxlen=TRAIL_MAX_POINTS)  # Just (x, y)
prev_center = None
no_detection_streak = 0
frame_idx = 0

print("Processing video with SMOOTH & INSTANT Hawk-Eye trail...")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    frame_idx += 1
    display_frame = frame.copy()

    # === Instant detection with your model ===
    results = model.track(
        source=frame,
        persist=True,
        tracker="bytetrack.yaml",
        conf=MIN_CONF,
        classes=0,
        verbose=False
    )[0]

    current_center = None

    # === Get best ball detection ===
    if results.boxes is not None and len(results.boxes) > 0:
        boxes = results.boxes.xyxy.cpu().numpy()
        confs = results.boxes.conf.cpu().numpy()

        best_conf = 0
        best_cx, best_cy = None, None

        for box, conf in zip(boxes, confs):
            x1, y1, x2, y2 = box
            size = (x2 - x1 + y2 - y1) / 2
            if not (MIN_SIZE <= size <= MAX_SIZE):
                continue
            cx = int((x1 + x2) / 2)
            cy = int((y1 + y2) / 2)

            if conf > best_conf:
                best_conf = conf
                best_cx, best_cy = cx, cy

        if best_cx is not None:
            current_center = (best_cx, best_cy)

    # === Instantly update trail (zero lag) ===
    if current_center:
        cx, cy = current_center

        if prev_center is not None:
            dist = np.hypot(cx - prev_center[0], cy - prev_center[1])
            if dist < MAX_JUMP_PER_FRAME or no_detection_streak < MAX_GAP_FRAMES:
                # Add point immediately
                trail_points.append((cx, cy))
                prev_center = (cx, cy)
                no_detection_streak = 0
            # else: ignore outlier
        else:
            trail_points.append((cx, cy))
            prev_center = (cx, cy)
            no_detection_streak = 0
    else:
        no_detection_streak += 1

    # === Draw ULTRA-SMOOTH glowing trail ===
    if len(trail_points) >= 2:
        points_array = np.array(trail_points, dtype=np.int32)

        # Draw multiple layered glowing lines for smoothness and depth
        for thickness, color, alpha in [
            (28, (120, 255, 255), 0.4),   # Outer soft glow
            (18, (50, 255, 220), 0.6),     # Mid glow
            (10, (0, 255, 180), 0.8),      # Bright cyan
            (6,  (255, 255, 255), 1.0),    # White hot core
        ]:
            cv2.polylines(
                display_frame,
                [points_array],
                isClosed=False,
                color=color,
                thickness=thickness,
                lineType=cv2.LINE_AA  # Anti-aliased = super smooth!
            )

        # Current ball - bright instant highlight
        latest_x, latest_y = trail_points[-1]
        cv2.circle(display_frame, (latest_x, latest_y), 22, (0, 255, 255), -1, cv2.LINE_AA)
        cv2.circle(display_frame, (latest_x, latest_y), 22, (255, 255, 255), 6, cv2.LINE_AA)
        cv2.circle(display_frame, (latest_x, latest_y), 12, (255, 255, 255), -1, cv2.LINE_AA)

    # === Clean professional text ===
    cv2.putText(display_frame, "HAWK-EYE", (w//2 - 180, 80),
                cv2.FONT_HERSHEY_DUPLEX, 2.5, (255, 255, 255), 7, cv2.LINE_AA)
    cv2.putText(display_frame, "HAWK-EYE", (w//2 - 178, 80),
                cv2.FONT_HERSHEY_DUPLEX, 2.5, (0, 240, 255), 4, cv2.LINE_AA)

    cv2.putText(display_frame, f"Trail: {len(trail_points)} points", (30, h - 30),
                cv2.FONT_HERSHEY_SIMPLEX, 1.1, (0, 255, 100), 3, cv2.LINE_AA)

    # Write frame immediately
    out.write(display_frame)

# Cleanup
cap.release()
out.release()
cv2.destroyAllWindows()

print("\nDONE! Ultra-smooth, instant Hawk-Eye trajectory created!")
print(f"Saved: {output_video}")
print("Trail now appears IMMEDIATELY and looks buttery smooth with anti-aliased glow.")
```
